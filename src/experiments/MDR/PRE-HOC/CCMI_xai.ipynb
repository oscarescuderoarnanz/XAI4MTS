```python
import os
import pandas as pd
import pickle
# Define the filenames and output names
split = [1, 2, 3]  
split_directory = './Results-App1/'

for s in split:
    # Load data
    results_pop = load_from_pickle(os.path.join(split_directory, f"CCMI_results_population_s{s}.pkl"))
    results_df = results_pop[f'results_df_{s}'].copy()  

    # Process data
    results_df["Base_Feature"] = results_df["Feature"].str.rsplit("_", n=1).str[0]
    results_df["Time_Step"] = results_df["Feature"].str.rsplit("_", n=1).str[1].astype(int)

    # Aggregate
    df_grouped = results_df.groupby(["Base_Feature", "Time_Step"])["MI Value"].sum().reset_index()

    # Convert to matrix
    df_pivot = df_grouped.pivot(index="Base_Feature", columns="Time_Step", values="MI Value").fillna(0)
    df_pivot.columns = [f"T_{col}" for col in df_pivot.columns]

    # Convert to long format
    df = df_pivot.reset_index()
    df_long = df.melt(id_vars=["Base_Feature"], var_name="Time_Step", value_name="MI Value")
    df_long = df_long.drop(columns=["Time_Step"]).rename(columns={"Base_Feature": "Feature"})

    # Plot the results
    plot_filename = f"../Figures/PRE-HOC/CCMI_s{s}.pdf"
    plot_dataframe(df_long, plot_filename)
```

```python
split_directory = './Results-App1/'
results_pop1 = load_from_pickle(os.path.join(split_directory, f"CCMI_results_population_s1.pkl"))
results_pop2 = load_from_pickle(os.path.join(split_directory, f"CCMI_results_population_s2.pkl"))
results_pop3 = load_from_pickle(os.path.join(split_directory, f"CCMI_results_population_s3.pkl"))

# Extract dataframes
dfs = [results_pop1['results_df_1'], results_pop2['results_df_2'], results_pop3['results_df_3']]

# Process each dataframe to create 71x14 matrices
matrices = []
for df in dfs:
    df[['feature_name', 'time_step']] = df['Feature'].str.rsplit('_', n=1, expand=True)
    df['time_step'] = df['time_step'].astype(int)
    df = df.groupby(['feature_name', 'time_step'], as_index=False)['MI Value'].mean()  # Handles duplicates
    matrix = df.pivot(index='feature_name', columns='time_step', values='MI Value')
    matrix = matrix.sort_index(axis=1)
    print(np.median(matrix))
    matrices.append(matrix)

# Define the correct order of features
desired_order = [
            'AMG', 'CAR', 'CF1', 'CF3', 'CF4',
            'Others', 'GLI', 'LIN', 'LIP', 'MAC', 'NTI', 'OXA', 'PAP', 'PEN', 'POL',
            'QUI', 'SUL', 'hoursVM', 'acinet.$_{pc}$', 'enterobac.$_{pc}$', 'enteroc.$_{pc}$',
            'pseud.$_{pc}$', 'staph.$_{pc}$', 'others.$_{pc}$', 'hoursICU',
            '# pat_atb', '# pat_MR',
            'CAR.$_{n}$', 'PAP.$_{n}$', 'Falta.$_{n}$', 'QUI.$_{n}$',
            'OXA.$_{n}$', 'PEN.$_{n}$', 'CF3.$_{n}$', 'GLI.$_{n}$',
            'CF4.$_{n}$', 'SUL.$_{n}$', 'NTI.$_{n}$', 'LIN.$_{n}$',
            'AMG.$_{n}$', 'MAC.$_{n}$', 'CF1.$_{n}$', 'POL.$_{n}$',
            'LIP.$_{n}$', '# pat_ttl','posture.$_{change}$',
            'insulin', 'nutr_art', 'sedation', 'relax', 'hep_fail',
            'renal_fail', 'coag_fail', 'hemo_fail',
            'resp_fail', 'multi_fail', 'n_transf',
            'vasoactive.$_{drug}$', 'dosis_nems', 'hoursTracheo', 'hoursUlcer',
            'hoursHemo', 'C01 PIVC 1',
            'C01 PIVC 2', 'C02 CVC - YD',
            'C02 CVC - SD', 'C02 CVC - SI', 'C02 CVC - FD',
            'C02 CVC - YI', 'C02 CVC - FI', '# catheters'
]

# Ensure only features present in all matrices and in desired order
filtered_order = [feat for feat in desired_order]

# Reorder matrices according to the filtered order
matrices = [m.loc[filtered_order] for m in matrices]

# Convert matrices to numpy arrays for voting
matrices_np = [m.to_numpy() for m in matrices]

# Majority voting: if MI > 0 in at least two matrices, set to 1; otherwise, 0
majority_matrix = (
    ((matrices_np[0] >= 0.000000000000000010010707824421618).astype(int) + 
     (matrices_np[1] >= 0.000000000000000032034265038149174).astype(int) + 
     (matrices_np[2] >= 0.0).astype(int)) >= 2
).astype(int)
```

```python
import os
import numpy as np
import shutil

# Directories
input_dir = "../../../../DATA/MDR"
output_dir = os.path.join(input_dir, "FS_CCMI")
os.makedirs(output_dir, exist_ok=True)

# Subfolders (e.g., s1, s2, s3) in the input directory
subfolders = ["s1", "s2", "s3"]

# Process each subfolder
for subfolder in subfolders:
    subfolder_path = os.path.join(input_dir, subfolder)
    
    if os.path.exists(subfolder_path):
        print(f"Processing folder: {subfolder}")
        
        # Create a specific folder for each subfolder in the output directory
        subfolder_output_dir = os.path.join(output_dir, subfolder)
        os.makedirs(subfolder_output_dir, exist_ok=True)
        
        # Process train and validation files
        data_types = ["train", "val"]
        for data_type in data_types:
            for i in range(5):  # Looping from 0 to 4 for different versions
                file_name = f"X_{data_type}_tensor_{i}robustNorm.npy"
                file_path = os.path.join(subfolder_path, file_name)
                
                if os.path.exists(file_path):
                    print(f"Processing: {file_name}")
                    data = np.load(file_path)  # Shape (P, 14, 71)
                    
                    # Apply the majority voting mask
                    mask = np.tile(majority_matrix.T, (data.shape[0], 1, 1))  # Broadcast mask to match shape
                    data_filtered = np.where(mask == 1, data, 666)  # Apply the mask
                    
                    # Save the modified data
                    output_file = os.path.join(subfolder_output_dir, file_name)
                    np.save(output_file, data_filtered)
                    print(f"Saved: {output_file}")
                else:
                    print(f"Not found: {file_name}")

        # Process X_test_tensor_robustNorm.npy
        test_file_name = "X_test_tensor_robustNorm.npy"
        test_file_path = os.path.join(subfolder_path, test_file_name)

        if os.path.exists(test_file_path):
            print(f"Processing: {test_file_name}")
            test_data = np.load(test_file_path)  # Shape (P, 14, 71)

            # Apply the majority voting mask
            mask = np.tile(majority_matrix.T, (test_data.shape[0], 1, 1))  # Broadcast mask to match shape
            test_data_filtered = np.where(mask == 1, test_data, 666)  # Apply the mask
            
            # Save the modified test data
            test_output_file = os.path.join(subfolder_output_dir, test_file_name)
            np.save(test_output_file, test_data_filtered)
            print(f"Saved: {test_output_file}")
        else:
            print(f"Not found: {test_file_name}")

        # Copy y_train, y_val, y_test (CSV files)
        y_data_types = ["train", "val", "test"]
        for data_type in y_data_types:
            if data_type == "test":
                y_file_name = "y_test_tensor_robustNorm.csv"
                y_file_path = os.path.join(subfolder_path, y_file_name)
                
                if os.path.exists(y_file_path):
                    print(f"Copying: {y_file_name}")
                    output_y_file = os.path.join(subfolder_output_dir, y_file_name)
                    shutil.copy(y_file_path, output_y_file)
                    print(f"Copied: {output_y_file}")
                else:
                    print(f"Not found: {y_file_name}")
            else:
                for i in range(5):  # Looping from 0 to 4 for train and val
                    y_file_name = f"y_{data_type}_tensor_{i}robustNorm.csv"
                    y_file_path = os.path.join(subfolder_path, y_file_name)
                    
                    if os.path.exists(y_file_path):
                        print(f"Copying: {y_file_name}")
                        output_y_file = os.path.join(subfolder_output_dir, y_file_name)
                        shutil.copy(y_file_path, output_y_file)
                        print(f"Copied: {output_y_file}")
                    else:
                        print(f"Not found: {y_file_name}")
    else:
        print(f"Subfolder not found: {subfolder}")

print("Processing completed.")
```
